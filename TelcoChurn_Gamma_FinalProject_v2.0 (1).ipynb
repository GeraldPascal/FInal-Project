# **1. Introduction**

## **1.1 Context**
<blockquote style="text-align: justify">
Telecommunication companies are now facing significant challenges in retaining customers amid increasingly fierce market competition. One of the most critical issues is customer churn, which occurs when customers stop using their services. Losing customers not only impacts revenue but also incurs substantial costs in acquiring new ones. In this context, the analyzed Telco company has historical customer data, including demographic information, services used, and payment behaviors. Leveraging this data, a comprehensive analysis was conducted to identify churn patterns and develop a predictive model to estimate the likelihood of churn in the future.
</blockquote>

## **1.2 Problem Statement**
<blockquote style="text-align: justify">
In the highly competitive telecommunications industry, customer retention has become a business priority that is equally — if not more — important than acquiring new customers. The cost of acquiring a new customer can be up to five times higher than the cost of retaining an existing one. Therefore, companies need to implement data-driven strategies to identify customers at risk of churn before it happens. Unfortunately, many companies still struggle with recognizing the early signs that a customer is about to churn, identifying the key factors that contribute most to a customer’s decision to leave, and implementing the right interventions quickly and efficiently. This is where the role of predictive analytics and machine learning becomes crucial.
</blockquote>

## **1.3 Goals**
<blockquote style="text-align: justify">

1. **Churn Understanding**  
   Identify the key factors that influence customer churn through exploratory analysis and predictive modeling.

2. **Customer Segmentation**  
   Cluster customers based on service usage, contract types, and payment behavior to uncover distinct personas and risk profiles.

3. **Revenue Optimization**  
   Quantify the financial impact of churn and explore how customer lifetime value (CLTV) is affected across segments.

4. **Operational Improvement**  
   Evaluate how services, billing methods, and contract structures correlate with churn, identifying levers for operational efficiency.

5. **Building a Machine Learning Model**
   Building an accurate and interpretable machine learning–based churn prediction model.

6. **Providing Business Insights**
   Building an accurate and interpretable machine learning–based churn prediction model.
</blockquote>

## **1.4 Analytical Approach**
<blockquote style="text-align: justify">

- Data understanding & cleaning
- Perform exploratory data analysis (EDA) to identify patterns and correlations.
- Apply clustering techniques to define customer personas.
- Develop a churn prediction model to identify high-risk customers.
- Generate business-driven recommendations supported by visual insights.
</blockquote>

## **1.5 Evaluation Metrics**
<blockquote style="text-align: justify">

- Accuracy – the proportion of correct predictions out of all predictions.
- Precision (macro) – measures the accuracy of positive predictions for both classes equally.
- Recall (macro) – focuses on minimizing false negatives in the churn class, aligned with the business objective.
- F1-score (macro) – the balance between precision and recall.
- ROC-AUC – the model’s ability to distinguish between churn and non-churn classes.
</blockquote>

`OVERALL DEMOGRAPHIC TREND`

```python
# Plot overall demographic trends
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle('Overall Demographic Trends', fontsize=16, fontweight='bold')
demo_features = ['gender', 'SeniorCitizen', 'Partner', 'Dependents']

for ax, feature in zip(axes.flat, demo_features):
    counts = df[feature].value_counts()
    total = counts.sum()
    sns.barplot(x=counts.index, y=counts.values, palette='Set2', ax=ax)
    for i, val in enumerate(counts.values):
        percent = 100 * val / total
        ax.annotate(f'{percent:.1f}%', (i, val + total*0.02), ha='center', va='bottom', fontsize=9)
    ax.set_title(feature)
    ax.set_ylabel('Count')
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.savefig(os.path.join(output_dir, "3.1 demographics_overall_trend.png"))
plt.show()
```

`MULTI-BAR PLOT (Countplot with hue)`

```python
# Function to annotate count on top of bars
def annotate_count(ax):
    for p in ax.patches:
        height = p.get_height()
        if height == 0:
            continue
        ax.annotate(f'{int(height)}', 
                    (p.get_x() + p.get_width() / 2., height), 
                    ha='center', va='bottom', fontsize=9, color='black')

# Plot setup
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Customer Demographics and Churn Breakdown', fontsize=16, fontweight='bold')

plot_configs = [
    ('gender', axes[0, 0], 'Gender vs Churn'),
    ('SeniorCitizen', axes[0, 1], 'Senior Citizen vs Churn'),
    ('Partner', axes[1, 0], 'Partner vs Churn'),
    ('Dependents', axes[1, 1], 'Dependents vs Churn'),
]

for feature, ax, title in plot_configs:
    sns.countplot(data=df, x=feature, hue='Churn', palette='Set2', ax=ax)
    ax.set_title(title)
    annotate_count(ax)

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.savefig(os.path.join(output_dir, "3.1 demographics_churn_breakdown.png"))
plt.show()
```

`HORIZONTAL STACKED PLOT (Churn Rate %)`

```python
# Plot churn rate by demographic groups
fig, axes = plt.subplots(1, 4, figsize=(20, 5))
fig.suptitle('Churn Rate (%) by Demographic Groups', fontsize=16, fontweight='bold')

for ax, feature in zip(axes, demo_features):
    churn_rate = df[df['Churn'] == 'Yes'][feature].value_counts() / df[feature].value_counts() * 100
    churn_rate = churn_rate.sort_values(ascending=True)
    sns.barplot(x=churn_rate.values, y=churn_rate.index, palette='Blues_r', ax=ax)
    for i, val in enumerate(churn_rate.values):
        ax.text(val + 1, i, f'{val:.1f}%', va='center', fontsize=9)
    ax.set_title(f'{feature}', fontsize=12)
    ax.set_xlabel('Churn Rate (%)')
    ax.set_xlim(0, max(churn_rate.values) + 10)

plt.tight_layout(rect=[0, 0, 1, 0.92])
plt.savefig(os.path.join(output_dir, "3.1 churn_rate_by_demographics.png"))
plt.show()
```

### **Key Insights from Visuals**

1. **Gender**
   - The distribution of churn is similar between males and females.
   - Churn rates:  
     - **Female:** 27.0%  
     - **Male:** 26.2%  
   - Gender does **not** appear to be a major churn driver.

2. **Senior Citizen**
   - Senior citizens have a **much higher churn rate** than non-senior customers.
   - Churn rates:  
     - **SeniorCitizen = Yes:** 41.7%  
     - **SeniorCitizen = No:** 23.7%
   - Being a senior citizen is **positively correlated** with churn.

3. **Partner**
   - Customers **without a partner** churn at a higher rate.
   - Churn rates:  
     - **Partner = No:** 33.0%  
     - **Partner = Yes:** 19.7%
   - Not having a partner is a **risk factor** for churn.

4. **Dependents**
   - Customers **without dependents** churn more frequently.
   - Churn rates:  
     - **Dependents = No:** 31.3%  
     - **Dependents = Yes:** 15.5%
   - Having dependents is **negatively correlated** with churn (protective factor).

### **Business Interpretation**

- **Targeted retention efforts** may be most effective among senior citizens, single customers, and those without dependents.
- Gender-based retention programs are likely unnecessary.
- Demographic segmentation can improve targeting of offers and communications for churn prevention.

`PLOTTING WITH ANNOTATED SEGMENTS`

```python
def plot_segmented_distribution(feature, bins, labels, color, title, xlabel, filename, custom_labels=False, value_format="{:.0f}"):
    plt.figure(figsize=(10,5))
    sns.histplot(df[feature], bins=50, kde=True, color=color)
    for b in bins:
        plt.axvline(b, color='gray', linestyle='--', linewidth=1)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel('Count')
    # Annotate ranges or custom labels
    for i in range(len(bins)-1):
        if labels is not None:
            text = (
                f"{labels[i]}:\n{value_format.format(bins[i]+0.01 if i>0 else bins[i])}–{value_format.format(bins[i+1])}"
            )
        else:
            text = f"{value_format.format(bins[i]+0.01 if i>0 else bins[i])}–{value_format.format(bins[i+1])}"
        plt.text((bins[i]+bins[i+1])/2, plt.ylim()[1]*0.9,
                 text,
                 ha='center', fontsize=10, color='black', bbox=dict(facecolor='white', alpha=0.6, boxstyle='round'))
    plt.tight_layout()
    plt.savefig(f"../output/img/{filename}")
    plt.show()
```

# **4. Feature Engineering**

```python
# --- Segment 1: Feature Engineering ---
import numpy as np
import pandas as pd

def feature_engineering(df):
    df = df.copy()
    df['AvgMonthlyCost'] = df['TotalCharges'] / df['tenure'].replace(0, np.nan)
    df['TenureGroup'] = pd.cut(
        df['tenure'],
        bins=[0, 12, 36, 72],
        labels=['ShortTerm', 'MidTerm', 'LongTerm'],
        right=False
    )
    service_cols = [
        'PhoneService', 'MultipleLines', 'InternetService', 'OnlineSecurity',
        'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies'
    ]
    df['TotalServices'] = df[service_cols].apply(lambda x: np.sum(x == 'Yes'), axis=1)
    df['HasStreaming'] = ((df['StreamingTV'] == 'Yes') | (df['StreamingMovies'] == 'Yes')).astype(int)
    df['IsHighSpender'] = (df['MonthlyCharges'] > df['MonthlyCharges'].median()).astype(int)
    df['PaperlessAndElectronic'] = (
        (df['PaperlessBilling'] == 'Yes') & (df['PaymentMethod'].str.contains('electronic', case=False))
    ).astype(int)
    df['HasPartnerOrDependents'] = (
        (df['Partner'] == 'Yes') | (df['Dependents'] == 'Yes')
    ).astype(int)
    df['IsNewCustomer'] = (df['tenure'] < 6).astype(int)
    return df
```

### **Feature Engineering**

In this stage, the process involves creating or modifying features from the original data so they are more representative for the model.

From the notebook contents, the steps taken include:
- Converting categorical features into numerical form (label encoding or one-hot encoding).
- Creating derived variables that better represent the data, for example:
    - Combining rare categories into a single “Other” category.
    - Converting “Yes/No” variables into binary values (1/0).
- Adjusting data types (integer, float, category) as needed.

The purpose of these steps is to enable the machine learning model to process the data more optimally, as most algorithms cannot directly handle string data.

### **Actionable Insights**

- This process helps reduce dimensionality (too many features) when rare categories are merged.
- Binary encoding for “Yes/No” columns prevents the dummy variable trap.

- Proper feature engineering can improve model accuracy by providing a more suitable representation of variable relationships.

# **5. Preprocessing**

```python
# --- Segment 2: Preprocessing ---
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler, FunctionTransformer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Column lists after feature engineering
numerical_cols = [
    'tenure', 'MonthlyCharges', 'TotalCharges',
    'AvgMonthlyCost', 'TotalServices'
]
categorical_cols = [
    'gender', 'SeniorCitizen', 'Partner', 'Dependents', 'PhoneService',
    'MultipleLines', 'InternetService', 'OnlineSecurity', 'OnlineBackup',
    'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies',
    'Contract', 'PaperlessBilling', 'PaymentMethod', 'TenureGroup'
]
binary_cols = [
    'HasStreaming', 'IsHighSpender', 'PaperlessAndElectronic',
    'HasPartnerOrDependents', 'IsNewCustomer'
]

target_col = 'Churn'
y = df[target_col].map({'No': 0, 'Yes': 1})

raw_columns = [
    'gender', 'SeniorCitizen', 'Partner', 'Dependents', 'tenure', 'PhoneService',
    'MultipleLines', 'InternetService', 'OnlineSecurity', 'OnlineBackup',
    'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies',
    'Contract', 'PaperlessBilling', 'PaymentMethod', 'MonthlyCharges', 'TotalCharges'
]
X = df[raw_columns]

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# Full preprocessing pipeline (feature engineering + scaling + encoding)
preprocessor = Pipeline([
    ('feature_engineering', FunctionTransformer(feature_engineering)),
    ('col_transform', ColumnTransformer([
        ('num', StandardScaler(), numerical_cols + binary_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
    ]))
])
```

# **6. Model Benchmarking**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import StratifiedKFold, cross_val_score, train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier
from catboost import CatBoostClassifier
from sklearn.ensemble import GradientBoostingClassifier
from lightgbm import LGBMClassifier
from sklearn.ensemble import ExtraTreesClassifier

from sklearn.metrics import (
    make_scorer, precision_score, recall_score, f1_score, roc_auc_score
)

# 4. Define models
models = {
    'RandomForest': RandomForestClassifier(class_weight='balanced'),
    'ExtraTrees': ExtraTreesClassifier(class_weight='balanced'),
    'LightGBM': LGBMClassifier(class_weight='balanced', verbose=-1),
    'XGBoost': XGBClassifier(scale_pos_weight=(y_train == 0).sum() / (y_train == 1).sum(), verbosity=0),
    'CatBoost': CatBoostClassifier(auto_class_weights='Balanced', verbose=0)
}



# 5. Scoring dictionary
scoring = {
    'accuracy': 'accuracy',
    'precision': make_scorer(precision_score, average='macro', zero_division=0),
    'recall': make_scorer(recall_score, average='macro', zero_division=0),
    'f1': make_scorer(f1_score, average='macro', zero_division=0),
    'roc_auc': 'roc_auc'
}

# 6. Cross-validation setup (on train data only)
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# 7. Collect all per-fold scores
fold_scores = {metric: {} for metric in scoring.keys()}

for name, model in models.items():
    pipeline = Pipeline(steps=[
        ('feature_engineering', FunctionTransformer(feature_engineering)),
        ('preprocessor', preprocessor),
        ('classifier', model)
    ])
    for metric in scoring.keys():
        try:
            scores = cross_val_score(
                pipeline, X_train, y_train, cv=cv, scoring=scoring[metric]
            )
        except Exception as e:
            print(f"{name} failed for metric {metric}: {e}")
            scores = np.array([np.nan]*cv.get_n_splits())
        fold_scores[metric][name] = scores

# (Optional) You can now use X_test, y_test for your final, blind evaluation after model selection.
```

```python
import matplotlib.patches as mpatches

# 9. Visualization
metrics_to_plot = ['accuracy', 'precision', 'recall', 'f1', 'roc_auc']
for metric in metrics_to_plot:
    plt.figure(figsize=(12, 6))
    all_scores = np.concatenate([fold_scores[metric][name] for name in fold_scores[metric]])
    ymin = max(0, np.nanmin(all_scores) - 0.01)
    ymax = min(1, np.nanmax(all_scores) + 0.01)
    
    # Compute average per model and find the max
    model_avg_scores = {name: np.nanmean(fold_scores[metric][name]) for name in fold_scores[metric]}
    max_avg = max(model_avg_scores.values())
    best_models = [name for name, avg in model_avg_scores.items() if np.isclose(avg, max_avg)]
    
    # Store plot handles and labels for custom legend
    handles, labels = [], []
    for name in fold_scores[metric]:
        scores = fold_scores[metric][name]
        avg_score = model_avg_scores[name]
        # Plot and save the line handle
        line, = plt.plot(
            range(1, len(scores) + 1),
            scores,
            marker='o',
            label=f'{name} (avg={avg_score:.3f})'
        )
        plt.hlines(
            y=avg_score,
            xmin=1, xmax=len(scores),
            colors='gray', linestyles='dashed'
        )
        handles.append(line)
        labels.append(f'{name} (avg={avg_score:.3f})')
        
    # Custom legend with green background for the best model(s)
    legend = plt.legend(handles=handles, labels=labels, frameon=True)
    for idx, text in enumerate(legend.get_texts()):
        if any(best_name in text.get_text() for best_name in best_models):
            text.set_bbox(dict(facecolor='limegreen', alpha=0.5, edgecolor='none'))
    plt.title(f'{metric.capitalize()} per Fold for Each Model')
    plt.xlabel('Fold')
    plt.ylabel(metric.capitalize())
    plt.ylim(ymin, ymax)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()



```
